```python
import pandas as pd
github_raw_url='https://raw.githubusercontent.com/LeowYongSheng/ADALL_github/refs/heads/main/laptop_prices_2024_sgd_TL.csv'
try:
  df=pd.read_csv(github_raw_url)
  print("successfully loaded data from Github")
  display(df.head())
except Exceptionn as e:
  print("Error loading data: (e)")
  print("Please ensure the URL is correct and the file format is compatible with 'pd.read_csv',")
```

Basic Exploratory Data Analysis

### 1. Data Structure Overview

Let's start by getting a general overview of the dataset, including column names, data types, and non-null values.

```python
print(df.info())
```

### 2. Descriptive Statistics for Numerical Columns

This will provide insights into the distribution of numerical data, such as mean, standard deviation, min, max, and quartiles.

```python
display(df.describe())
```

### 3. Check for Missing Values

Identifying missing values is crucial for data quality. We'll see how many missing values each column has.

```python
missing_values = df.isnull().sum()
print('Missing values per column:\n', missing_values[missing_values > 0])

if missing_values.sum() == 0:
    print('No missing values found in the dataset.')
```

### 4. Check for Duplicate Rows

Duplicate rows can skew analysis, so it's good practice to identify and handle them.

```python
duplicates = df.duplicated().sum()
print(f'Number of duplicate rows: {duplicates}')

if duplicates > 0:
    print('Consider removing duplicate rows if they represent redundant entries.')
```

### 5. Unique Values and Counts for Categorical Columns

For categorical features, it's useful to see their unique values and their frequencies to catch inconsistencies or potential encoding issues.

Let's look at `Brand`, `CPU`, `RAM`, `Storage`, `GPU` for now.

```python
categorical_cols = ['Brand', 'CPU', 'Storage_Type', 'GPU']
for col in categorical_cols:
    if col in df.columns:
        print(f'\nUnique values and counts for {col}:')
        display(df[col].value_counts().head(10)) # Display top 10 for brevity
    else:
        print(f'Column {col} not found in DataFrame.')
```

### 6. Data Quality Suggestions and Next Steps

Based on the EDA, here are some general suggestions to improve data quality and prepare the dataset for analysis:

*   **Missing Values**: If there are missing values (e.g., in `Storage_Type`, `GPU_Manufacturer`), decide on an imputation strategy (e.g., mean, median, mode, or a constant value) or consider dropping rows/columns if missingness is extensive.
*   **Data Types**: Ensure columns like `Rating` and `Number_of_Ratings` are numeric. If they appear as `object` (string), they might contain non-numeric characters that need cleaning.
*   **Categorical Encoding**: Convert categorical features (e.g., `Brand`, `CPU`, `RAM`, `Storage`, `GPU`) into numerical representations using techniques like One-Hot Encoding or Label Encoding, especially if planning to use machine learning models.
*   **Feature Engineering**: Consider creating new features from existing ones. For instance, extracting numerical values from `RAM` (e.g., '16GB' -> 16) or `Storage` (e.g., '512GB SSD' -> 512, 'SSD').
*   **Outliers**: Investigate potential outliers in numerical columns like `Price_SGD`, `Rating`, `Number_of_Ratings` which might require transformation or removal.
*   **Unit Consistency**: Ensure all units are consistent across columns (e.g., if 'GB' and 'TB' are used for storage, convert 'TB' to 'GB').
